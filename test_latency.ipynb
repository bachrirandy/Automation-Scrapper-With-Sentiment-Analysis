```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca file CSV
file_path = 'hasil_scraping_sumberdaya.csv'
df = pd.read_csv(file_path)

# Mengatur style visualisasi
sns.set_style("whitegrid")

# --- 1. Visualisasi untuk Bab 1.1.1 (Analisis Latency) ---

# Bar chart untuk status keberhasilan (dengan perbaikan)
plt.figure(figsize=(8, 6))
status_counts = df['Status'].value_counts()
sns.barplot(
    x=status_counts.index, 
    y=status_counts.values, 
    palette='viridis', 
    hue=status_counts.index,  # Ditambahkan sesuai saran
    legend=False              # Ditambahkan sesuai saran
)
plt.title('Jumlah Status Keberhasilan Scraping')
plt.xlabel('Status')
plt.ylabel('Jumlah')
plt.savefig('2visualisasi_status_keberhasilan111.png')
plt.close()

# Box plot untuk perbandingan waktu respon (dengan perbaikan)
plt.figure(figsize=(10, 7))
sns.boxplot(
    x='Status', 
    y='Waktu_Respon_detik', 
    data=df, 
    palette='coolwarm',
    hue='Status',             # Ditambahkan sesuai saran
    legend=False              # Ditambahkan sesuai saran
)
plt.title('Perbandingan Waktu Respon (Latency) Berdasarkan Status')
plt.xlabel('Status')
plt.ylabel('Waktu Respon (detik)')
plt.savefig('2visualisasi_perbandingan_latency111.png')
plt.close()

# --- 2. Visualisasi untuk Bab 1.1.2 (Analisis Sumber Daya) ---
# (Kode untuk bagian ini tidak menghasilkan warning, jadi tetap sama)

plt.figure(figsize=(14, 8))
plt.subplot(2, 1, 1)
plt.plot(df.index, df['CPU_Usage_persen'], marker='.', linestyle='-', color='b')
plt.title('Penggunaan CPU per Permintaan Scraping')
plt.xlabel('Urutan Permintaan (Request)')
plt.ylabel('Penggunaan CPU (%)')

plt.subplot(2, 1, 2)
plt.plot(df.index, df['Memory_Usage_MB'], marker='.', linestyle='-', color='g')
plt.title('Penggunaan Memori per Permintaan Scraping')
plt.xlabel('Urutan Permintaan (Request)')
plt.ylabel('Penggunaan Memori (MB)')
plt.tight_layout()
plt.savefig('2visualisasi_sumberdaya111.png')
plt.close()

plt.figure(figsize=(8, 6))
sns.histplot(df['Memory_Usage_MB'], kde=True, bins=20, color='forestgreen')
plt.title('Distribusi Penggunaan Memori')
plt.xlabel('Penggunaan Memori (MB)')
plt.ylabel('Frekuensi')
plt.savefig('2visualisasi_distribusi_memori111.png')
plt.close()

print("Kode yang diperbarui siap digunakan. Ini akan menghasilkan visualisasi yang sama tanpa menampilkan FutureWarning.")
```
